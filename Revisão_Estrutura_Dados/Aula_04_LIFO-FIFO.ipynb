# Para Lembrar

```python
# Criando uma pilha vazia
pilha = []

# Empilhando elementos (push)
pilha.append(10)
pilha.append(20)
pilha.append(30)

print("Pilha atual:", pilha)

# Removendo o elemento do topo (pop)
topo = pilha.pop()
print("Elemento removido:", topo)
print("Pilha após o pop:", pilha)
```

.pop == tirar elemento de cima;

.append == adiciona um elemento no topo.

#Exercício 01 -- O simulador "Desfazer"
Objetivo : Implentar uma pilha para gerenciar ações em um editor de texto.

Requisitos : Armazene cada ação (digitar, apagar, substituir). Ao acionar a função desfazer, o programa deve remover e reverter estritamente o último comando inserido.

```python
cod = 0
def adicionar(pilha):
    print("\n=== ADICIONAR ===")
    def codigo():
        global cod
        cod += 1
        return cod
    def nome_input():
        while True:
            nome = input("Digite o nome a ser adicionado: ")
            if nome.isalpha():
                return nome
            else:
                print("O nome não deve conter espaços ou números!")
    nome_temporario = nome_input()
    decisao = input("Deseja confirmar o cadastro deste nome? (S/N)\n")
    if decisao.upper() == 'S':
        novo_codigo = codigo()
        digitar = [novo_codigo, nome_temporario]
        pilha.append(digitar)
        print(f"Nome cadastrado com sucesso!\nCódigo é **{novo_codigo}**\n")
    elif decisao.upper() == 'N':
        print("OK! Ação cancelada e nome não salvo!\n")
    else:
        print("Digite uma resposta válida!\nS para SIM / N para NÃO\n")

# def apagar(pilha):
# def substituir(pilha):

def mostrar(pilha):
  print("\n=== MOSTRAR ===")
  if not pilha:
    print("Não há nomes adicionados!\n")
    return
  for i, digitar in enumerate(pilha):
    print(f"Código : {digitar[0]}")
    print(f"Nome : {digitar[1]}")
    print("---------------------")
  print("\n")


def menu():
  pilha = []
  while True:
    print("=== MENU ===")
    print("1 - Adicionar")
    print("2 - Apagar")
    print("3 - Subsituir")
    print("4 - Mostrar")
    print("0 - Sair")
    try:
      escolha = int(input("Escolha : "))
    except ValueError:
      print("Entrada inválida. Digite um número para a opção.")
      continue
    if escolha == 0:
      print("OK! Saindo do programa...")
      break
    elif escolha == 1:
      adicionar(pilha)
    # elif escolha == 2:
    #   apagar(pilha)
    # elif escolha == 3:
    #   substituir(pilha)
    elif escolha == 4:
      mostrar(pilha)
    else:
      print("Digite uma resposta válida!")

menu()
```
