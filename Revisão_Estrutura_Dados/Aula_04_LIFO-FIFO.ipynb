#Exercício 01 -- O simulador "Desfazer"
Objetivo : Implentar uma pilha para gerenciar ações em um editor de texto.

Requisitos : Armazene cada ação (digitar, apagar, substituir). Ao acionar a função desfazer, o programa deve remover e reverter estritamente o último comando inserido.

```python
cod = 0

def get_valid_name_input():
    while True:
        nome = input("Digite o nome: ").title()
        if nome.replace(' ', '').isalpha(): # Allow spaces but check if it's alphabetic otherwise
            return nome
        else:
            print("O nome não deve conter números ou caracteres especiais!\n")

def adicionar(pilha):
    print("\n=== ADICIONAR ===")
    def codigo():
        global cod
        cod += 1
        return cod
    nome_temporario = get_valid_name_input()
    decisao = input(f"Deseja confirmar o cadastro do nome '{nome_temporario}'? (S/N)\n")
    if decisao.upper() == 'S':
        novo_codigo = codigo()
        digitar = [novo_codigo, nome_temporario]
        pilha.append(digitar)
        print(f"Nome cadastrado com sucesso!\nCódigo é **{novo_codigo}**\n")
    elif decisao.upper() == 'N':
        print("OK! Ação cancelada e nome não salvo!\n")
    else:
        print("Digite uma resposta válida!\nS para SIM / N para NÃO\n")

def apagar(pilha):
  print("\n=== APAGAR ===")
  if not pilha:
    print("Não há nomes para apagar. A pilha está vazia!\n")
    return

  ultimo_item = pilha[-1]
  nome_do_topo = ultimo_item[1]

  resposta_apagar = input(f"Deseja apagar o último nome adicionado, '{nome_do_topo}'? (S/N)\n").upper()

  if resposta_apagar == 'S':
    resposta_confirmar = input("Tem certeza que deseja apagar permanentemente? (S/N)\n").upper()
    if resposta_confirmar == 'S':
      removido = pilha.pop()
      print(f"Nome '{removido[1]}' com código '{removido[0]}' removido com sucesso!\n")
    elif resposta_confirmar == 'N':
      print("OK! Ação cancelada e nome não foi removido (confirmação negada)!\n")
    else:
      print("Entrada inválida para confirmação. Ação cancelada.\n")
  elif resposta_apagar == 'N':
    print("OK! Ação cancelada e nome não foi removido!\n")
  else:
    print("Digite uma resposta válida para apagar (S/N). Ação cancelada.\n")

def substituir(pilha):
  print("\n=== SUBSTITUIR ===")
  if not pilha:
    print("Não há nomes para editar!\n")
    return

  nome_para_editar = input("Digite o nome a ser editado : ").title()
  found = False
  for i, item in enumerate(pilha):
    if nome_para_editar == item[1]: # Compare title-cased input with stored name
      print(f"Nome '{item[1]}' encontrado!")
      confirmacao_substituir = input(f"Deseja substituir '{item[1]}'? (S/N)\n").upper()
      if confirmacao_substituir == 'S':
        print(f"Digite o novo nome para '{item[1]}' :")
        novo_nome = get_valid_name_input()
        pilha[i] = [
            item[0], # Keep the original code
            novo_nome
        ]
        print(f"Nome atualizado para '{pilha[i][1]}' com código '{pilha[i][0]}'.\n")
        found = True
        break # Exit loop after finding and updating the first match
      elif confirmacao_substituir == 'N':
          print(f"Substituição de '{item[1]}' cancelada.\n")
          found = True # Mark as found to avoid 'not found' message
          break
      else:
          print("Entrada inválida para confirmação. Substituição cancelada.\n")
          found = True # Mark as found to avoid 'not found' message
          break
  if not found:
      print(f"Nome '{nome_para_editar}' não encontrado na pilha.\n")

def mostrar(pilha):
  print("\n=== MOSTRAR ===")
  if not pilha:
    print("Não há nomes para mostrar!\n")
    return
  for i, item in enumerate(pilha):
    print(f"Código : {item[0]}")
    print(f"Nome : {item[1]}")
    print("---------------------")
  print("\n")


def menu():
  pilha = []
  while True:
    print("=== MENU ===")
    print("1 - Adicionar")
    print("2 - Apagar")
    print("3 - Substituir")
    print("4 - Mostrar")
    print("0 - Sair")
    try:
      escolha = int(input("Escolha : "))
    except ValueError:
      print("Entrada inválida. Digite um número para a opção.\n")
      continue
    if escolha == 0:
      print("OK! Saindo do programa...")
      break
    elif escolha == 1:
      adicionar(pilha)
    elif escolha == 2:
      apagar(pilha)
    elif escolha == 3:
      substituir(pilha)
    elif escolha == 4:
      mostrar(pilha)
    else:
      print("Digite uma resposta válida!\n")

menu()
```
