##Atividade 1 — Listas
Crie um programa que:
1. crie uma lista com 10 números inteiros;
2. solicite os valores ao usuário;
3. exiba todos os números;
4. calcule a soma;
5. calcule a média.
###Desafio
Além disso, encontre:
- o maior valor;
- o menor valor;
- a quantidade de números pares.

```python
lista = []
num = []

while len(lista) < 10:
    numero_digitado = int(input("Digite um número inteiro pra sua lista : "))
    lista.append(numero_digitado)
    if numero_digitado % 2 == 0:
        num.append(numero_digitado)

print("\nTodos os números:", " ".join(map(str, lista)))

soma = sum(lista)
print("A soma dos elementos é : ", soma)

maiornum = max(lista)
print("O maior número da lista é : ", maiornum)

menornum = min(lista)
print("O menor número da lista é : ", menornum)

print("Os números pares são :", " ".join(map(str, num)))
```

##Atividade 2 — Matrizes
Crie uma matriz 3 × 3.

Solicite os valores ao usuário e:
1. exiba a matriz;
2. calcule a soma de todos os elementos;
3. calcule a soma da diagonal principal;
4. encontre o maior elemento.

```python
linhas = 3
colunas = 3
matriz = [ ]

for i in range(linhas):
  linha = [ ]
  for j in range(colunas):
    valor = int(input(f"Digite a posição {i}, {j}: "))
    linha.append(valor)
  matriz.append(linha)

print("\nMatriz digitada:")
for linha in matriz:
  print(linha)

soma_total = 0
soma_diagonal = 0
maior_elemento = matriz[0][0]

for i in range(linhas):
  for j in range(colunas):
    soma_total += matriz[i][j]

    if matriz[i][j] > maior_elemento:
      maior_elemento = matriz[i][j]

    if i == j:
      soma_diagonal += matriz[i][j]

print(f"\nSoma de todos os elementos: {soma_total}")
print(f"Soma da diagonal principal: {soma_diagonal}")
print(f"Maior elemento da matriz: {maior_elemento}")
```

Crie uma dataclass:

0 - Sair

```python
def cadastrar_produto(estoque):
  ultimo_codigo = 0
  print("\n======= CADASTRAR NOVO PRODUTO =======")
  def codigo_produto():
    global ultimo_codigo
    ultimo_codigo += 1
    return ultimo_codigo
  def nome_produto():
    while True:
      nomeProd = input("Digite o nome do produto : ").strip().title()
      if nomeProd.isalpha():
        return nomeProd
      else:
        print("Erro! Digite apenas letras, sem números ou espaços.")
  def preco_produto():
    while True:
      try:
        precoProd = float(input("Digite o preco do produto com duas casas decimais : "))
        if precoProd >= 0:
          return precoProd
        else:
          print("O preço não pode ser negativo.")
      except ValueError:
        print("Entrada inválida. Digite um número para o preço.")
  def qtd_produto():
    while True:
      try:
        qtdProd = int(input("Digite a quantidade de produtos : "))
        if qtdProd >= 0:
          return qtdProd
        else:
          print("A quantidade não pode ser negativa.")
      except ValueError:
        print("Entrada inválida. Digite um número inteiro para a quantidade.")


  produto = [codigo_produto(), nome_produto(), preco_produto(), qtd_produto()]
  estoque.append(produto)
  print("Produto cadastrado com sucesso!!")
  print(f"O código do seu produto é {produto[0]}\n")

def listar_produtos(estoque):
  qtdEst = len(estoque)
  print(f"\n======= LISTAR PRODUTOS -- VOCÊ TEM {qtdEst} PRODUTOS NO ESTOQUE =======")
  if not estoque:
    print("Estoque vazio!!\n")
    return
  for i, produto in enumerate(estoque):
    print(f"--- Produto {i+1} ---")
    print(f"Código : {produto[0]}")
    print(f"Nome : {produto[1]}")
    print(f"Preço : R${produto[2]:.2f}")
    print(f"Quantidade : {produto[3]}\n")


def buscar_produto(estoque):
  print("\n======= BUSCAR PRODUTOS =======")
  codigo_busca = input("Digite o código do produto que busca : ")
  encontrados = [e for e in estoque if codigo_busca in e[0].title()]
  if codigo_busca:
    for e in estoque:
      print("\n======= Produto encontrado!! =======")
      print(f"Códogi : {e[0]}")
      print(f"Nome : {e[1]}")
      print(f"Preço : R${e[2]}")
      print(f"Quantidade : {e[3]}\n")
  else:
    print("Produto não encontrado... Tente novamente!\n")
    return

# def valor_estoque(estoque):
# def produto_caro(estoque):

def menu():
  estoque = []
  while True:
    print("===== SISTEMA DE ESTOQUE =====")
    print("1 - Cadastrar Produto")
    print("2 - Listar Produtos")
    print("3 - Buscar Produto")
    print("4 - Valor Total no Estoque")
    print("5 - Produto Mais Caro")
    print("0 - Sair")
    try:
      escolha = int(input("Escolha : "))
    except ValueError:
      print("Entrada inválida. Digite um número para a opção.")
      continue
    if escolha == 0:
      print("\nOK! Saindo do sistema...")
      break
    elif escolha == 1:
      cadastrar_produto(estoque)
    elif escolha == 2:
      listar_produtos(estoque)
    elif escolha == 3:
      buscar_produto(estoque)
    # elif escolha == 4:
    #   valor_estoque(estoque)
    # elif escolha == 5:
    #   produto_caro(estoque)
    else:
      print("Digite um valor válido!!")

menu()
```
