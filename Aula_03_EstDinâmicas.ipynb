#Atividade 02
Crie:

n1 = Node("A")

n2 = Node("B")

n3 = Node("C")

Conecte:

A → B → C → None

Depois percorra a estrutura e produza:

A
B
C

```python
class Node:
    def __init__(self, dado):
        self.dado = dado
        self.proximo = None

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3
n3.proximo = None
print(n1.dado)
print(n1.proximo.dado)
print(n1.proximo.proximo.dado)
```
