```python
# 1. Representação de um Nó na Árvore Binária de Busca (ABB)
class No:
    def __init__(self, valor):
        # Dado armazenado no nó
        self.valor = valor
        # Referência para o filho da esquerda (valores menores)
        self.esquerda = None
        # Referência para o filho da direita (valores maiores)
        self.direita = None


# 2. Função de Inserção Guiada na ABB (com base nas regras da aula)
def inserir(raiz, valor):
    # Se a árvore (ou subárvore) estiver vazia, cria um novo nó
    if raiz is None:
        return No(valor)

    # Valores menores seguem para a esquerda
    if valor < raiz.valor:
        raiz.esquerda = inserir(raiz.esquerda, valor)
    # Valores maiores seguem para a direita
    elif valor > raiz.valor:
        raiz.direita = inserir(raiz.direita, valor)

    # Retorna o nó (raiz atual) sem alterações se o valor já existir
    return raiz


# 3. Função de Busca Iterativa (conforme o slide 14 da aula)
def buscar(raiz, alvo):
    # Ponteiro 'atual' começa na raiz da árvore
    atual = raiz

    # Percorre o caminho até encontrar o nó ou chegar a uma posição vazia (None)
    while atual is not None:
        # Se o valor do nó atual for igual ao alvo, o elemento foi encontrado
        if alvo == atual.valor:
            return atual

        # Decisão: se o alvo for menor, avança para a esquerda
        if alvo < atual.valor:
            atual = atual.esquerda
        # Caso contrário (se for maior), avança para a direita
        else:
            atual = atual.direita

    # Se saiu do laço while sem encontrar o valor, retorna None
    return None


# -------------------------------------------------------------
# 4. Execução do Teste Rápido (Slide 15)
# -------------------------------------------------------------
if __name__ == "__main__":
    # Sequência do teste rápido: 50, 30, 70, 20, 40, 60, 80
    valores = [50, 30, 70, 20, 40, 60, 80]

    raiz = None
    for v in valores:
        raiz = inserir(raiz, v)

    print("=== Respostas do Teste Rápido ===")
    print(f"• Raiz da árvore: {raiz.valor}")

    # Teste de busca 1: Procurando pelo valor 60
    resultado_60 = buscar(raiz, 60)
    print(
        f"• Busca por 60: {'Encontrado (' + str(resultado_60.valor) + ')' if resultado_60 else 'Não encontrado'}"
    )
    print("  Caminho percorrido até 60: 50 -> 70 -> 60")

    # Teste de busca 2: Procurando pelo valor 65 (Inexistente)
    resultado_65 = buscar(raiz, 65)
    print(
        f"• Busca por 65: {'Encontrado (' + str(resultado_65.valor) + ')' if resultado_65 else 'Não encontrado (None)'}"
    )
    print("  Caminho percorrido até 65: 50 -> 70 -> 60 -> (direita é None)")
```
